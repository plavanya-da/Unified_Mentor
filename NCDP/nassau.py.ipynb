```python
Product_Sales=data.groupby('Product Name')['Sales'].sum().sort_values(ascending=False)
print(Product_Sales)
```

```python
data['Revenue Contribution']=data['Product Name'].map(Product_Sales)/data['Sales'].sum() * 100
print(data)
```

```python
Product_Profit= data.groupby('Product Name')['Gross Profit'].sum()
print(Product_Profit)
```

```python
data['Profit Contribution']=data['Product Name'].map(Product_Profit)/data['Gross Profit'].sum()*100
print(data)
```

```python
Margin_Volatility=data.groupby('Product Name')['Gross Margin (%)'].std()
print(Margin_Volatility)
data['Margin Volatility'] = data['Product Name'].map(Margin_Volatility)
data
data= data.drop('Margin Volatility',axis=1)
data
```

```python
data['Order Date'] = pd.to_datetime(data['Order Date'], dayfirst=True)
data['Month'] = data['Order Date'].dt.to_period('M')
```

```python
data
```

```python
Monthly_Margin=data.groupby(['Product Name','Month'])['Gross Margin (%)'].mean()
print(Monthly_Margin)
```

```python
margin_volatility= Monthly_Margin.groupby('Product Name').std()
print(margin_volatility)
```

```python
data['Margin Volatility']=data['Product Name'].map(margin_volatility)
data
data=data.drop('Margin Volatility',axis=1)
data
```

```python
High_rank=data.groupby('Product Name')['Gross Profit'].sum()
data['Profit_rank']=data['Product Name'].map(High_rank.rank(ascending=False))
data
```

```python
margin_rank=data.groupby('Product Name')['Gross Margin (%)'].sum()
data['Margin_rank']=data['Product Name'].map(margin_rank.rank(ascending=False))
data
```

```python
High_Profit_Products=data.groupby('Product Name')['Gross Profit'].sum().sort_values(ascending=False)
High_Sales_Products=data.groupby('Product Name')['Sales'].sum().sort_values(ascending=False)
Low_Sales_Products=data.groupby('Product Name')['Sales'].sum().sort_values(ascending=True)
Low_Profit_Products = data.groupby('Product Name')['Gross Profit'].sum().sort_values(ascending=True)
High_Margin_Products = data.groupby('Product Name')['Gross Margin (%)'].mean().sort_values(ascending=False)
Low_Margin_Products = data.groupby('Product Name')['Gross Margin (%)'].mean().sort_values(ascending=True)
```

```python
high_profit_margin = set(High_Profit_Products.head(10).index) & set(High_Margin_Products.head(10).index)
print(high_profit_margin)
```

```python
high_sales_low_margin=set(High_Sales_Products.head(10).index) & set(Low_Margin_Products.head(10).index)
print(high_sales_low_margin)
```

```python
Low_Sales_Low_Profit=set(Low_Sales_Products.head(10).index) & set(Low_Profit_Products.head(10).index)
print(Low_Sales_Low_Profit)
```

```python
print("High Profit + High Margin:", high_profit_margin)
print("High Sales + Low Margin:", high_sales_low_margin)
print("Low Sales + Low Profit:", Low_Sales_Low_Profit)
```
